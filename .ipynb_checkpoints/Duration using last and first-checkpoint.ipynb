```python
import rasterio
import numpy as np

def calculate_duration(year):
    # Paths to the input files
    first_file = f'G:/Hangkai/Anttarctic Vegetation Dynamic/Version_2_data/Antarctica_Vegetation_Duration/First_last/First_{year}.tif'
    last_file = f'G:/Hangkai/Anttarctic Vegetation Dynamic/Version_2_data/Antarctica_Vegetation_Duration/First_last/Last_{year}.tif'
    
    # Output file path
    duration_file = f'G:/Hangkai/Anttarctic Vegetation Dynamic/Version_2_data/Antarctica_Vegetation_Duration/First_last/Duration_{year}.tif'
    
    # Open the first and last files
    with rasterio.open(first_file) as first_src, rasterio.open(last_file) as last_src:
        # Read the rasters as arrays
        first_arr = first_src.read(1)
        last_arr = last_src.read(1)
        
        # Create a mask for valid data
        valid_mask = (first_arr != 0) & (first_arr != 9999) & (~np.isnan(first_arr)) & (first_arr != first_src.nodata) & \
                     (last_arr != 0) & (last_arr != 9999) & (~np.isnan(last_arr)) & (last_arr != last_src.nodata)
        
        # Initialize the duration array with zeros
        duration_arr = np.zeros_like(first_arr, dtype=np.float32)
        
        # Calculate the duration for valid data points
        duration_arr[valid_mask] = np.where(last_arr[valid_mask] >= first_arr[valid_mask],
                                            last_arr[valid_mask] - first_arr[valid_mask] + 1,
                                            (last_arr[valid_mask] + 365) - first_arr[valid_mask] + 1)

        # Write the results to a new file with the same metadata as the first file
        with rasterio.open(duration_file, 'w', **first_src.meta) as dst:
            dst.write(duration_arr, 1)

# Loop through each year from 2002 to 2023
for year in range(2002, 2024):
    calculate_duration(year)
    print(f"Duration for {year} calculated and saved.")
```

```python
import rasterio
import numpy as np
from scipy.stats import linregress
import matplotlib.pyplot as plt

def calculate_mean_greater_than_zero(file_path):
    with rasterio.open(file_path) as data_src:
        data = data_src.read(1)
    masked_data = np.ma.masked_less_equal(data, 0)
    return masked_data.mean()

years = range(2002, 2024)
means = []

for year in years:
    data_file = f'G:/Hangkai/Anttarctic Vegetation Dynamic/Version_2_data/Antarctica_Vegetation_Duration/First_last/Land/Duration_{year}.tif'
    mean_value = calculate_mean_greater_than_zero(data_file)
    means.append(mean_value)
    print(f"Year: {year}, Mean: {mean_value}")

# Perform linear regression
slope, intercept, r_value, p_value, std_err = linregress(years, means)

# Print regression results
print(f"Slope: {slope}, Intercept: {intercept}, R-squared: {r_value**2}, p-value: {p_value}")

# Plot the results
plt.figure(figsize=(10, 6))
plt.plot(years, means, 'o', label='Mean values > 0')
plt.plot(years, intercept + slope * np.array(years), 'r', label='Linear fit')
plt.xlabel('Year')
plt.ylabel('Mean Value of Data > 0')
plt.title('Trend of Mean Values from 2002 to 2023')
plt.legend()
plt.grid(True)
plt.show()
```
