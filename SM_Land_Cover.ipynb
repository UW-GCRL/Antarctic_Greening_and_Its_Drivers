```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

# === Load CSV ===
csv_path = r"G:/Hangkai/Anttarctic Vegetation Dynamic/DART/Antarctic_NDVI_all/Antarctic_NDVI_all/Data/LandCover/TableB/Snow_10_Algae_10_Moss_10_Lichen_10/lc_map.csv"
df = pd.read_csv(csv_path, header=None)

# Convert to numpy array
lc = df.values

# === Define colors for each class ===
# 0 = snow, 1 = rock, 2 = moss, 3 = algae, 4 = lichen
colors = [
    "#FFFFFF",  # snow
    "#7f7f7f",  # rock (gray)
    "#2ca02c",  # moss (green)
    "#17becf",  # algae (cyan / blue-green)
    "#bcbd22"   # lichen (yellow)
]

cmap = ListedColormap(colors)

# === Plot ===
plt.figure(figsize=(10, 10), dpi=600)
plt.imshow(lc, cmap=cmap, interpolation='nearest')
plt.axis("off")

# === Legend ===
import matplotlib.patches as mpatches
legend_labels = [
    ("0 = snow", colors[0]),
    ("1 = rock", colors[1]),
    ("2 = moss", colors[2]),
    ("3 = algae", colors[3]),
    ("4 = lichen", colors[4])
]

handles = [mpatches.Patch(color=c, label=lab) for lab, c in legend_labels]

plt.legend(
    handles=handles,
    loc="lower left",
    fontsize=12,
    frameon=True,
    borderpad=1
)

plt.tight_layout()
plt.savefig(r"C:\Users\hyou34\OneDrive - UW-Madison\manuscripts\paper\Greening Antarctica\nature_version\Supplementary_Information\High_Resolution_Figures\SM_Simulated_Land_Cover.png", dpi=600, bbox_inches="tight")

plt.show()
```
